```python
import pandas as pd
from pathlib import Path
```

```python
project_root = Path.cwd().parent
bronze_data_path = project_root / "data"/"bronze"/"agricultural_production.parquet"
```

```python
data = pd.read_parquet(bronze_data_path)
data.head()
```

```python
data.info()
```

```python
data['area_plantada_ha'].groupby(data['ano']).sum()
```

```python
data['area_plantada_ha'].groupby(data['cultura']).sum()
```

```python
data['area_plantada_ha'].groupby(data['estado']).sum()
```

```python

verificacao_espacos = data.groupby(data['cultura'].str.strip())['cultura'].unique()

espacos_duplicados = verificacao_espacos[verificacao_espacos.str.len() > 1]

espacos_duplicados
```

```python

verificacao_estado = data.groupby(data['estado'].str.upper())['estado'].unique()

estados_com_duplicidade = verificacao_estado[verificacao_estado.str.len() > 1]

estados_com_duplicidade
```

```python

verificacao_municipio = data.groupby(data['municipio'].str.upper())['municipio'].unique()

cidades_com_duplicidade = verificacao_municipio[verificacao_municipio.str.len() > 1]

cidades_com_duplicidade
```
